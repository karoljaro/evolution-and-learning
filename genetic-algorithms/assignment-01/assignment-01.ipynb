## Zadanie 1 - Losowa populacja początkowa

**Cel:** Utworzyć funkcję, która dla zadanej liczby chromosomów (osobników) oraz długości 
chromosomu (liczby bitów  ) tworzy losową populację początkową.

```python
def create_population(population_size, chromosome_length, rng=None):
    """Zwróć losową populację binarną o zadanych wymiarach.

    population_size: dodatnia całkowita liczba osobników.
    chromosome_length: dodatnia całkowita liczba bitów na osobnika.
    rng: opcjonalny generator NumPy pozwalający powtórzyć eksperyment.
    """
    for name, value in (
        ("population_size", population_size),
        ("chromosome_length", chromosome_length),
    ):
        if isinstance(value, (bool, np.bool_)) or not isinstance(value, (int, np.integer)):
            raise TypeError(f"{name} musi być liczbą całkowitą.")
        if value <= 0:
            raise ValueError(f"{name} musi być większe od zera.")

    if rng is None:
        rng = np.random.default_rng()


    return rng.integers(
        low=0,
        high=2,
        size=(population_size, chromosome_length),
        dtype=np.uint8,
    )
```

```python
rng = np.random.default_rng(42)

population = create_population(
    population_size=5,
    chromosome_length=8,
    rng=rng,
)

print("Populacja początkowa:")
print(population)
print("Wymiary:", population.shape)
print("Wylosowane wartości:", np.unique(population))
```

### Wnioski

Otrzymamy 5 osobników, gdzie każdy chromosom ma 8 bitów. Funkcja zwraca wyłącznie wartości binarne i pozwala zmieniać oba wymiary populacji.

Losowanie nie gwarantuje różnych chromosomów ani równej liczby zer i jedynek. Powtórzenia są dopuszczalne. Ponowne utworzenie generatora z tym samym ziarnem odtwarza wynik; kolejne wywołanie z już użytym generatorem kontynuuje losowanie.
